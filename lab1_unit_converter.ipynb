## Overview — thinking before typing
### What is computational thinking?

Programmers rarely start by typing. They start by thinking about the problem in four ways, and you will use all four in this lab:

* __Decomposition__ — break the problem into smaller pieces. *Get a number, do the arithmetic, show the answer.*
* __Pattern recognition__ — notice what repeats. *Every conversion in this lab has the same shape: read, calculate, print.*
* __Abstraction__ — ignore what does not matter. *The converter does not care whether 100 degrees is a fever or a boiling kettle. It only needs the number.*
* __Algorithm design__ — put the steps in an order a computer can follow, with no gaps.

### <span style="color:crimson">Variables and expressions</span>
A __variable__ is a name that holds a value. The `=` sign means *store this value under this name* — it does not mean "equals" the way it does in algebra.

An __expression__ is anything Python can calculate to produce a value, like `distance_miles * 1.609344`.

````python
# a variable is a name that holds a value
distance_miles = 26.2          # the length of a marathon
distance_km = distance_miles * 1.609344
print(distance_km)
````

Python follows the usual order of operations: `*` and `/` happen before `+` and `-`. When in doubt, add parentheses — they cost nothing and make your intent obvious.

````python
print(2 + 3 * 4)      # multiplication first  -> 14
print((2 + 3) * 4)    # parentheses first     -> 20
````

### <span style="color:crimson">The four core data types</span>
Every value in Python has a __type__. In this course you will use four constantly:

| Type | What it holds | Example |
|---|---|---|
| `int` | whole numbers | `42` |
| `float` | numbers with a decimal point | `98.6` |
| `str` | text, written inside quotes | `"Chicago"` |
| `bool` | exactly two values | `True` or `False` |

The built-in `type()` tells you the type of any value.

````python
print(type(42))
print(type(98.6))
print(type("Chicago"))
print(type(True))
````

Two surprises catch almost everyone. A number inside quotes is text, not a number. And `/` always produces a `float`, even when the division comes out even.

````python
print(type(42))
print(type("42"))     # quotes make it text
print(10 / 2)         # 5.0, not 5
print(type(10 / 2))
````

### <span style="color:crimson">Type conversion</span>
You can convert a value from one type to another with `int()`, `float()` and `str()`.

<span style="color:crimson">__Watch out:__</span> `int()` does not round. It cuts the decimal off completely. If you want rounding, use `round()`.

````python
print(float(7))         # 7.0
print(int("42") + 1)    # 43   -> text converted to a number
print(str(42) + "!")    # 42!  -> a number converted to text
print(int(3.9))         # 3    -> NOT 4. int() cuts off the decimal
print(round(3.9))       # 4    -> round() is what rounds
````

Why this matters for your converter: __`input()` always gives you text.__ Even if the user types `100`, Python receives the string `"100"`. You must convert it before you can do arithmetic with it.

### <span style="color:crimson">Input and formatted output</span>
`input()` shows a prompt, waits for the user to type something, and gives you back what they typed — always as a `str`.

````python
city = input("Which city are you visiting? ")
print(f"Enjoy your trip to {city}!")
````

The `f` before the quotes makes an __f-string__: anything inside `{curly braces}` is replaced with its value. Add `:.2f` inside the braces to show exactly two decimal places.

````python
distance_km = 42.164802
print(distance_km)                             # too many decimals
print(f"{distance_km:.2f}")                    # exactly two decimals -> 42.16
print(f"A marathon is {distance_km:.2f} km.")
````
